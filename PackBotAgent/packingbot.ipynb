```python
# ── Flask Routes ───────────────────────────────────────────────────
app = Flask(__name__)

@app.route("/")
def index():
    return render_template_string(HTML_PAGE)


@app.route("/api/generate", methods=["POST"])
def api_generate():
    data = request.get_json(silent=True) or {}
    dest = data.get("destination", "").strip()
    if not dest:
        return jsonify(error="Destination is required."), 400

    user_prompt = f"""Trip Details:
- Destination: {dest}
- Dates: {data.get('dates', 'Not specified')} ({data.get('duration', 7)} days)
- Type: {data.get('trip_type', 'Leisure')}
- Activities: {data.get('activities', 'General sightseeing')}
- Accommodation: {data.get('accommodation', 'Hotel')}
- Transportation: {data.get('transport', 'Flight')}
- Luggage: {data.get('bags', 'Carry-on Only')}
- Gender presentation: {data.get('gender', 'Neutral')}
- Medical needs: {data.get('medical') or 'None'}
- Working remotely: {data.get('work', 'No')}

Generate the complete packing list following your output structure exactly."""

    try:
        response = ollama.chat(
            model="llama3.1:8b",
            messages=[
                {"role": "system", "content": SYSTEM_PROMPT},
                {"role": "user", "content": user_prompt},
            ],
            options={"temperature": 0.4},
        )
        packing_list = response["message"]["content"]
    except Exception as e:
        return jsonify(error=f"Ollama error: {e} — make sure `ollama serve` is running."), 502

    trip = {
        "destination": dest,
        "dates": data.get("dates", ""),
        "duration": data.get("duration", 7),
        "trip_type": data.get("trip_type", "Leisure"),
        "result": packing_list,
    }
    save_trip(trip)
    return jsonify(result=packing_list, id=trip["id"])


@app.route("/api/history")
def api_history():
    return jsonify(load_trips())


@app.route("/api/history/<int:trip_id>")
def api_history_detail(trip_id):
    for t in load_trips():
        if t.get("id") == trip_id:
            return jsonify(t)
    return jsonify(error="Trip not found."), 404
```
